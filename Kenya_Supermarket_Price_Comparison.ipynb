```python
import pandas as pd

data = {
    "Product": [
        "Sugar", "Sugar",
        "Rice", "Rice",
        "Maize Flour", "Maize Flour",
        "Cooking Oil", "Cooking Oil"
    ],
    "Brand": [
        "Brand A", "Brand A",
        "Brand B", "Brand B",
        "Brand C", "Brand C",
        "Brand D", "Brand D"
    ],
    "Package_Size": [
        "1 kg", "1 kg",
        "2 kg", "2 kg",
        "2 kg", "2 kg",
        "1 litre", "1 litre"
    ],
    "Retailer": [
        "Supermarket A", "Supermarket B",
        "Supermarket A", "Supermarket B",
        "Supermarket A", "Supermarket B",
        "Supermarket A", "Supermarket B"
    ],
    "Price_KSh": [
        180, 170,
        350, 330,
        160, 155,
        300, 285
    ]
}

df = pd.DataFrame(data)

df
```

```python
comparison = df.pivot(
    index=["Product", "Package_Size"],
    columns="Retailer",
    values="Price_KSh"
)

comparison["Price_Difference"] = (
    comparison["Supermarket A"]
    - comparison["Supermarket B"]
)

comparison["Cheaper_Supermarket"] = comparison.apply(
    lambda row: "Supermarket A"
    if row["Supermarket A"] < row["Supermarket B"]
    else "Supermarket B",
    axis=1
)

print(comparison)
```

```python
import matplotlib.pyplot as plt

comparison[["Supermarket A", "Supermarket B"]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Kenyan Supermarket Price Comparison")
plt.xlabel("Product")
plt.ylabel("Price (KSh)")
plt.xticks(rotation=45)
plt.legend(title="Retailer")
plt.tight_layout()
plt.show()
```

```python
# Check for missing values
print("Missing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("Duplicate rows:")
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Display the cleaned dataset
print("Cleaned dataset:")
print(df)
```

```python
average_prices = df.groupby("Retailer")["Price_KSh"].mean()

print("Average product price by supermarket:")
print(average_prices)
```

```python
# Save the cleaned dataset as a CSV file
df.to_csv("kenyan_supermarket_prices.csv", index=False)

print("Dataset saved successfully!")
```

KENYAN SUPERMARKET PRICE COMPARISON SYSTEM

1. Introduction

This project compares prices of selected household products in Kenyan supermarkets using Python.

2. Problem Statement

Consumers may find it difficult to compare prices across supermarkets. This project uses data analysis to identify price differences and lower-priced products.

3. Objectives

- To collect supermarket product prices.
- To clean and organize the data.
- To compare prices between supermarkets.
- To calculate average prices.
- To visualize the results using charts.

4. Tools Used

- Python
- Google Colab
- Pandas
- Matplotlib

5. Methodology

A sample dataset containing sugar, rice, maize flour, and cooking oil was created. Pandas was used to organize and clean the data, calculate averages, and compare prices. Matplotlib was used to create a bar chart.

6. Results and Findings

The average listed price was KSh 247.50 for Supermarket A and KSh 235.00 for Supermarket B. Supermarket B had lower sample prices for all four products. Rice had the largest price difference of KSh 20, while maize flour had the smallest difference of KSh 5.

These results are based on fictional sample prices, not verified current supermarket prices.

7. Conclusion

The project demonstrates how Python can be used to organize, analyze, and visualize supermarket prices to support price comparisons.

8. Recommendations

- Collect verified prices from permitted supermarket websites.
- Compare products of the same brand and package size.
- Record the date and source of each price.
- Update the dataset regularly.

9. Limitations

The dataset contains only eight sample records and uses fictional prices. Actual supermarket prices must be collected before drawing conclusions about the cheapest retailer in Kenya.
